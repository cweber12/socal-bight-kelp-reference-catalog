```python
%matplotlib inline
# No Software chunk in the PNG: matplotlib writes its own version into one by default, so
# the committed bytes would change on a matplotlib release with no pixel moving. The rest
# of the dict is the inline backend's default, which replacing it would otherwise drop.
%config InlineBackend.print_figure_kwargs = {"bbox_inches": "tight", "metadata": {"Software": None}}
import matplotlib.pyplot as plt

from kelpcatalog.notebook import load, provenance

# Every string on this figure is the file's own: the axis labels are its header row, the
# tick labels are YR values it states, and x is the order it lists its rows in.
text = load("noaa_oni")["oni.ascii.txt"].read_text(encoding="utf-8")
header, *rows = (line.split() for line in text.splitlines() if line.strip())

decades: dict[str, int] = {}
for position, row in enumerate(rows):
    if int(row[1]) % 10 == 0:
        decades.setdefault(row[1], position)

fig, ax = plt.subplots(figsize=(9, 3.2))
ax.plot(range(len(rows)), [float(row[3]) for row in rows], linewidth=0.9)
ax.set_xticks(list(decades.values()), labels=list(decades))
ax.set_xlim(0, len(rows) - 1)
ax.set_xlabel(header[1])
ax.set_ylabel(header[3])
fig.tight_layout()
# plt.show() before the caption, not after the cell: with the inline backend and no
# show(), matplotlib publishes the figure from a post_execute hook, after this cell's
# result - and the caption would commit above the plot it cites. Measured, #47.
plt.show()
provenance(sources=["noaa_oni"])
```
